# Programación dinámica

La programación dinámica o DP, es una **técnica de optimización** para resolver problemas complejos dividiéndolos en subproblemas más simples.

Esta técnica es especialmente útil cuando identificamos que la solución a un problema requiere resolver los mismos subproblemas múltiples veces.

Existen dos requisitos para aplicar programación dinámica:

1. **Subestructura óptima**: La solución óptima del problema puede construirse a partir de soluciones óptimas de sus subproblemas.

2. **Solapamiento de subproblemas**: El problema puede descomponerse en subproblemas que se repiten varias veces.

El primer paso para resolver un problema utilizando esta técnica, es descartar otras posibles soluciones. Los problemas que se resuelven con DP suelen ser similares a aquellos que se resuelven con, por ejemplo, un algoritmo **greedy**, donde basta con tomar decisión óptima en cada paso aislado para llegar a la solución óptima final. Sin embargo, en DP, la decisión óptima en cada paso depende de las decisiones tomadas en pasos anteriores.

## Definición del estado

El truco de la DP es almacenar los resultados de los subproblemas para evitar cálculos repetidos, generalmente mediante una lista o matriz `dp`.

Cada problema tiene ciertos parámetros que nos permiten definir un **estado** `dp[...]`. Un estado es una representación de un subproblema que nos permite identificarlo de manera única.

## Transición de estados

El siguiente paso es encontrar la fórmula matemática que nos permita calcular el valor de un estado desconocido, por ejemplo, `dp[i][x]`, a partir de otros estados ya calculados `dp[j][y]`.

> Generalmente cada subproblema puede abstraerse a una única decisión sencilla: realizar una acción o no realizarla.

Definir las transiciones también implica definir las restricciones de nuestros parámetros. La restricción más común es que las cantidades no pueden ser negativas.

La restricción más importante es solo evaluar hasta encontrar la respuesta al problema original. Por ejemplo, si nos piden Fibonacci de `n = 50`, no tendría sentido aplicar una restricción `0 <= i <= 51`, pues no necesitamos calcular Fibonacci de números mayores a 50.

> Estas restricciones nos permiten definir el tamaño de la tabla `dp`.
>
> Cada parámetro representa una dimensión de la tabla (generalmente son dos dimensiones, que se representan como filas y columnas).

## Fijación de casos base

Para resolver un subproblema, siempre necesitamos la solución a algún subproblema anterior, por lo que debemos definir manualmente la solución de al menos un subproblema, para no realizar cálculos infinitamente.

Para el ejemplo de Fibonacci, nuestros casos base son `F(0) = 0` y `F(1) = 1`, pues a partir de estos dos valores podemos calcular cualquier otro número de Fibonacci.

## Ubicación de la solución

Si pensamos `dp` como una tabla, debemos ubicar en qué celda de la misma estará la solución final. Muchas veces la solución final se encuentra en la última fila y última columna de la tabla, pero también puede ser, por ejemplo, el valor máximo de una fila de la tabla.

En el caso de Fibonacci, `dp` es una lista, y la solución final se encuentra en la última celda de la lista, `dp[n]`.

## Cálculo de complejidad espacial y temporal

Antes de empezar a programar, es importante calcular la complejidad temporal y espacial de nuestra solución. Si supera los límites de tiempo o memoria, sabemos que debemos realizar optimizaciones o buscar otra solución, sin perder tiempo programando.

### Complejidad espacial

Podemas calcular la complejidad espacial simplemente viendo el tamaño de la tabla `dp`, y el tipo de dato almacenado en cada celda.

> Por ejemplo, si tenemos una tabla de 1000 filas y 1000 columnas, que almacena valores de tipo `int` (4 bytes).
>
> $$
> 1000 \cdot 1000 \cdot 4 \space \text{bytes} = 4,000,000 \space \text{bytes} = 4 \space \text{MB}
> $$

Viendo nuestra fórmula de transición de estados, muchas veces podemos ver que un subproblema solo depende de un número limitado de subproblemas anteriores, por lo que podemos reducir la complejidad espacial almacenando únicamente los subproblemas necesarios para calcular el siguiente estado.

> Para calcular Fibonacci solo necesitamos los dos últimos números de la secuencia.
>
> $$
> F(n) = F(n - 1) + F(n - 2)
> $$
>
> Por lo tanto, no necesitamos una lista de `n` elementos, pues nos alcanza con almacenar únicamente los dos últimos resultados.

### Complejidad temporal

Calcular la complejidad temporal no siempre es tan sencillo como calcular la complejidad espacial. Dependiendo de las optimizaciones realizadas, la complejidad temporal puede ser mayor o menor a la complejidad espacial.

Si no realizamos optimizaciones, la complejidad temporal suele coincidir con el tamaño de la tabla `dp`.

Realizar optimizaciones de la complejidad espacial, como hicimos para Fibonacci, reduce el tamaño de la tabla a dos filas, pero aún debemos calcular Fibonacci para todos los números hasta `n`, es decir, `O(n)`.

Sin embargo, incluso en casos donde no es posible reducir el tamaño de la tabla, puede que no sea necesario calcular el subproblema correspondiente a cada una de las celdas, por lo que la complejidad temporal puede ser menor al tamaño de la tabla.

## Knapsack

_Knapsack_ es un problema clásico de programación dinámica, podemos encontrarlo en [atcoder.jp/contests/dp/tasks/dp_d](https://atcoder.jp/contests/dp/tasks/dp_d).

### 1. Definición del estado

Consideramos un parámetro `i`, que es la cantidad de elementos que consideramos, y `x`, que es la capacidad restante de la mochila.

`dp[i][x]` es el valor maximo de la mochila, considerando los primeros `i` elementos, con una capacidad `x`.

### 2. Transición de estados

Imaginemos que vamos aumentando el parámetro `i`, es decir, que vamos considerando un nuevo elemento en cada iteración. Para calcular `dp[i][x]` tenemos dos opciones:

- **No agarrar el nuevo elemento**: el resultado será el mismo que si no lo hubiéramos considerado.
  
  ```python
  dp[i - 1][x]
  ```

- **Agarrar el nuevo elemento**: el resultado será el valor del elemento que agarramos, más el mejor resultado que podemos obtener con la capacidad restante de la mochila, que es `x` menos el peso del elemento que acabamos de agarrar.
  
  ```python
  valor + dp[i - 1][x - peso]
  ```
  Utilizamos `i - 1` porque ya agarramos el elemento `i`, por lo que no podemos volver a considerarlo.

Debo quedarme con la mejor de las dos opciones (el máximo).

```python
dp[i][x] = max(
    dp[i - 1][x],
    valor + dp[i - 1][x - peso],
)
```

Como `i` y `x` son cantidades, no pueden ser negativas. Tampoco podemos considerar más elementos de los que tiene la lista (`n`), y no nos interesa evaluar una capacidad mayor a la que tiene la mochila de la consigna (`w`).

```python
0 <= i <= n
0 <= x <= w
```

### 3. Fijación de casos base

Si consideramos 0 elementos, el valor máximo de la mochila es 0. Es decir, para todo `x` se cumple `dp[0][x] = 0`.

```python
for x in range(w + 1):
    dp[0][x] = 0
```

Si la capacidad de la mochila es 0, el valor máximo de la mochila es 0. Es decir, para todo `i` se cumple `dp[i][0] = 0`.

```python
for i in range(n + 1):
    dp[i][0] = 0
```

### 4. Ubicación de la solución

Volviendo a la definición del estado, la solución al problema original es el valor máximo de la mochila considerando los `n` elementos (todos), para la capacidad `w` que nos da la consigna.

```python
res = dp[n][w]
```

### 5. Complejidad espacial y temporal

La complejidad espacial es el tamaño de la tabla `dp`, que tiene `w + 1` filas (va de `0` a `w`, la capacidad de la mochila), y `n + 1` columnas (va de `0` a `n`, la cantidad de elementos), y almacena números enteros (supongamos 4 bytes cada uno).

$$
O \left[ (w + 1)(n + 1)(4 \space \text{bytes}) \right] \approx O \left( w  \cdot n \cdot 4 \space \text{bytes} \right) =
\newline
O \left( 10^5 \cdot 100 \cdot 4 \space \text{bytes} \right) = O \left( 4 \cdot 10^7 \space \text{bytes} \right) =
\newline
O \left( 4 \cdot 10^4 \space \text{kilobytes} \right) = O \left( 4 \cdot 10^1 \space \text{megabytes} \right) =
\newline
O \left( 40 \space \text{megabytes} \right)
$$

Como no realizamos optimizaciones al tamaño de la tabla, la complejidad temporal es el número de celdas en la tabla.

$$
O \left[ (w + 1)(n + 1) \right] =
O \left( w \cdot n \right) =
O \left( 10^5 \cdot 100 \right) =
O \left( 10^7 \right)
$$

````python
elementos, capacidad = map(int, input().split())
pesos = []
valores = []

for _ in range(elementos):
    p, v = map(int, input().split())

    pesos.append(p)
    valores.append(v)

dp = [[0 for _ in range(capacidad + 1)] for _ in range(elementos + 1)]

for i in range(1, elementos + 1):
    for x in range(capacidad + 1):
        peso = pesos[i - 1]
        valor = valores[i - 1]

        if x >= peso:
            dp[i][x] = max(
                dp[i - 1][x],
                valor + dp[i - 1][x - peso],
            )
        else:
            dp[i][x] = dp[i - 1][x]

res = dp[elementos][capacidad]

print(res)
````
