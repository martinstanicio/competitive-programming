# Grafos

Un grafo es un conjunto de $V$ nodos (o vértices) y $E$ aristas que conectan pares de nodos.

````python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.Graph()

G.add_nodes_from(range(1, 6))
G.add_edges_from(
    [
        (1, 2),
        (1, 3),
        (1, 4),
        (2, 4),
        (2, 5),
        (3, 4),
        (4, 5),
    ]
)

plt.figure(figsize=(8, 4))
pos = nx.spring_layout(G, seed=3)
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="skyblue",
    node_size=2000,
    font_weight="bold",
)
plt.show()
````

## Grafos dirigidos y no dirigidos

Las aristas de un grafo pueden tener una única dirección (grafo dirigido) o no tener dirección (grafo no dirigido). Esto afecta cómo se pueden recorrer los nodos y cómo se representan las conexiones entre ellos.

## Grafos ponderados y no ponderados

Cada arista de un grafo puede tener asociado un peso, costo o distancia (grafo ponderado) o no tener ningún valor asociado (grafo no ponderado).

> Un grafo no ponderado puede representarse como un grafo ponderado donde todas las aristas tienen peso 1.

Un ejemplo de grafo ponderado es un mapa, donde cada ciudad es un nodo, y cada ruta entre ciudades es una arista con un peso que representa la distancia.

## Representación

Debemos encontrar una forma de representar un grafo en la memoria de la computadora, que nos permita realizar operaciones sobre él de manera eficiente.

### Lista de adyacencia

Por cada nodo del grafo, guardamos una lista de los nodos a los que está conectado. Es la representación más común.

```python
vecinos: list[list[int]]

vecinos[1] = [2, 3, 4]
vecinos[2] = [1, 4, 5]
vecinos[3] = [1, 4]
vecinos[4] = [1, 2, 3, 5]
vecinos[5] = [2, 4]
```

> Si las aristas tienen un peso asociado, podemos representarlas como tuplas de dos elementos: `(peso, nodo)`.

### Lista de aristas

Guardamos una lista de todas las aristas del grafo. Cada arista se representa como un par de nodos.

```python
aristas: list[tuple[int, int]]
aristas = [
    (1, 2),
    (1, 3),
    (1, 4),
    (2, 4),
    (2, 5),
    (3, 4),
    (4, 5),
]
```

> Si las aristas tienen un peso asociado, podemos representarlas como tuplas de tres elementos: `(nodo1, nodo2, peso)`.

## Árboles

Un árbol es un tipo especial de grafo que es **conexo** y **acíclico**. Esto significa que hay un camino entre cualquier par de nodos y no hay ciclos.

> Un árbol con $V$ nodos siempre tiene exactamente $V - 1$ aristas.

## Recorrido de grafos

Existen dos formas principales de recorrer todos los nodos de un grafo.

### DFS

La búsqueda en profundidad (Depth First Search) es un algoritmo que recorre un grafo explorando lo más profundo posible antes de retroceder. Es decir, si lo pensamos como un árbol, exploramos una única rama completa antes de pasar a la siguiente.

### BFS

La búsqueda en anchura (Breadth First Search) es un algoritmo que recorre un grafo explorando todos los nodos vecinos antes de pasar a los nodos del siguiente nivel. Es decir, si lo pensamos como un árbol, primero exploramos todas las ramificaciones a distancia uno, luego las de distancia dos, y así sucesivamente.

> BFS es la forma más sencilla de encontrar el camino de menor cantidad de aristas entre dos nodos de un grafo.
>
> Esto es lo mismo que decir que BFS encuentra el camino más corto en un grafo **no ponderado**.

## Dijkstra

Es un algoritmo que, dado un nodo inicial, nos permite encontrar la **distancia minima** a todos los demás nodos de un grafo **ponderado**.

> No podemos aplicar Dijkstra si el grafo tiene aristas con peso negativo.

## Bellman-Ford

Al igual que Dijkstra, Bellman-Ford nos permite obtener la **distancia mínima** desde un nodo inicial a todos los demás nodos de un grafo **ponderado**. Sin embargo, Bellman-Ford puede manejar **aristas con peso negativo**.

> Este algoritmo también nos permite detectar si un grafo tiene **ciclos negativos**.
>
> Si tenemos un ciclo negativo, es imposible encontrar una distancia mínima, ya que podríamos recorrer dicho ciclo negativo infinitas veces, y reducir la distancia a un nodo de manera indefinida.

## Floyd-Warshall

Si queremos encontrar la **distancia mínima entre cualquier par de nodos** de un grafo ponderado, podemos usar el algoritmo de Floyd-Warshall. Al igual que Bellman-Ford, soporta aristas con peso negativo, y permite detectar ciclos negativos.
