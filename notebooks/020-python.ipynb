## Enumerate

```python
enumerate(iterable, start=0)  # O(n)
```

Permite darle un índice a cada elemento de un iterable, devolviendo un objeto enumerado que contiene pares de índice y valor.

````python
precios = [100, 250, 80]

for i, p in enumerate(precios):  # (0, 100), (1, 250), (2, 80)
    print(f"Producto {i + 1}: ${p}")
````

## Zip

```python
zip(*iterables)  # O(n)
```

Dados dos o más iterables, combina el i-ésimo elemento de cada iterable en una tupla.

````python
precios = [100, 250, 80]
cantidades = [7, 6, 8]

for p, c in zip(precios, cantidades):  # (100, 7), (250, 6), (80, 8)
    print(f"{c} productos de ${p}")

total = sum(p * c for p, c in zip(precios, cantidades))  # expresión generadora
print(f"Total ${total}")
````

## Lista o vector

```python
list(iterable=None) -> list[T]  # O(n)
list[i] -> T  # O(1)

list.append(x: T) -> None  # O(1)
list.pop() -> T  # O(1)

list.pop(i: int) -> T  # O(n)
list.insert(i: int, x: T) -> None  # O(n)
```

Una lista de elementos, que nos permite leer, agregar y eliminar elementos de manera eficiente ($O(1)$), si se hace al final de la lista.

> En Python, la lista puede ser utilizada perfectamente como stack o pila, si solo utilizamos `append()` y `pop()`.

Insertar o eliminar elementos en posiciones arbitrarias de la lista es menos eficiente ($O(n)$), ya que requiere mover a todos los elementos posteriores al índice donde se inserta o elimina un elemento.

````python
def is_balanced(string: str) -> bool:
    pila = []

    for char in string:
        if char == "(":
            pila.append(char)  # guarda un paréntesis abierto en la pila
        elif char == ")":
            if not pila:  # si la pila está vacía, no hay paréntesis abierto para cerrar
                return False

            pila.pop()  # sino, saca un paréntesis abierto de la pila

    # si la pila queda vacía, todos los paréntesis están balanceados
    return len(pila) == 0


print(is_balanced("(())()"))
print(is_balanced("(()"))
print(is_balanced(")("))
````

## Queue / Stack (Cola / Pila)

```python
deque(iterable=None) -> deque[T]  # O(n)
deque[0] -> T  # O(1)
deque[-1] -> T  # O(1)
deque[i] -> T  # O(n)

deque.append(x: T) -> None  # O(1)
deque.pop() -> T  # O(1)

deque.appendleft(x: T) -> None  # O(1)
deque.popleft() -> T  # O(1)

deque.insert(i: int, x: T) -> None  # O(n)
deque.remove(value: T) -> None  # O(n)
```

Una deque (double-ended queue) actúa como pila y stack al mismo tiempo, permitiendo agregar y eliminar elementos de ambos extremos de manera eficiente ($O(1)$). Sin embargo, a diferencia de la lista, acceder a un elemento por índice es menos eficiente ($O(n)$), si no está en los extremos.

````python
from collections import deque

fila: deque[str] = deque()

fila.append("Cliente A")
fila.append("Cliente B")
fila.appendleft("VIP C")

while fila:
    persona = fila.popleft()
    print(f"Atendiendo a {persona}")
````

## Set (Conjunto)

```python
set(iterable=None) -> set[T]  # O(n)

set.add(x: T) -> None  # O(1)
set.remove(x: T) -> None  # O(1)

x in set -> bool  # O(1)
```

Similar a una lista, pero no permite elementos duplicados y no mantiene un orden de los elementos. Permite agregar, eliminar y buscar elementos de manera eficiente ($O(1)$).

````python
procesados: set[str] = set()

visitantes = ["40123456", "38999111", "40123456", "41555222"]

for dni in visitantes:
    procesados.add(dni)

if "40123456" in procesados:
    print("El DNI ya fue procesado.")

print(f"Se procesaron {len(procesados)} visitantes únicos.")
````

## Divmod

```python
divmod(a, b) -> tuple[int, int]  # O(1)
```

Nos permite obtener el cociente entero y el resto de una división en una sola operación, devolviendo una tupla con ambos valores. Es más eficiente que realizar ambas operaciones por separado.

````python
total_segundos = 135

# cociente, resto = divmod(dividendo, divisor)
minutos, segundos = divmod(total_segundos, 60)

print(f"{minutos} min {segundos} s")
````

## Itertools

```python
combinations(iterable, r)
pairwise(iterable)
permutations(iterable, r)
product(*iterables, repeat=1)
```

Itertools es un módulo de la biblioteca estándar de Python que nos permite realizar múltiples operaciones, como generar combinaciones, permutaciones, productos cartesianos, entre otras.

> Operaciones como calcular todas las permutaciones posibles de un conjunto son muy costosas ($O(n!)$), pero si fuera necesario para resolver un problema, es práctico utilizar `itertools` en vez de implementarlo manualmente.

````python
from itertools import combinations, permutations

jugadores = ["Martín", "Juan", "Nahuel"]

posibles_parejas = list(combinations(jugadores, 2))
posibles_resultados = list(permutations(jugadores))

print(f"{posibles_parejas = }")
print(f"{posibles_resultados = }")
````

````python
from itertools import pairwise

checkpoints = [0, 15, 35, 50]

distancia_entre_checkpoints = [b - a for a, b in pairwise(checkpoints)]

print(f"{distancia_entre_checkpoints = }")
````

````python
from itertools import product

entradas = ["Sopa", "Ensalada"]
principales = ["Carne", "Pasta", "Pizza"]

menus = list(product(entradas, principales))

print(f"{menus = }")
````

## Functools

```python
cache
lru_cache(maxsize=None)
```

Functools nos brinda dos decoradores que nos permiten almacenar en caché los resultados de funciones costosas, evitando cálculos repetidos.

`@cache` es un decorador que almacena en caché los resultados de una función, y `@lru_cache` (Least Recently Used) permite además establecer un límite de tamaño para la caché, guardando únicamente los `maxsize` elementos más recientes.

> Estos decoradores solo pueden ser aplicados a funciones que cumplan los siguientes requerimientos:
> 
> - **Determinismo:** para iguales entradas, la función debe devolver siempre el mismo resultado.
> - **Argumentos hashables:** los argumentos deben ser de tipos inmutables, como `int`, `float`, `str`, `tuple`, etc. Tipos mutables como `list`, `dict` o `set` no son hashables.

````python
from functools import cache


@cache
def fibonacci(n: int) -> int:
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(38))
````

### Uso de caché

Si un algoritmo no es aceptado por límites de tiempo, suele ser un indicador de que **no se está aplicando una solución óptima**, y probablemente sea necesario cambiar el enfoque. Muchas veces, si se utiliza cache en un algoritmo que supera el límite de tiempo, el algoritmo pasa a ser rechazado por superar el **límite de memoria**.

Sin embargo, en algunos casos con límites generosos, el uso de cache puede ser la diferencia entre escribir dos líneas de código, y tener que pensar en una solución completamente diferente.

> El parámetro `maxsize` de `@lru_cache` nos permite balancear el tiempo de ejecución y el uso de memoria.
> 
> Por ejemplo, si solo necesitamos calcular fibonacci de un único número, podemos establecer `maxsize=2`, para guardar únicamente los dos últimos resultados. De esta forma, minimizamos el uso de memoria y eliminamos los cálculos repetidos.
> 
> ```python
> @lru_cache(maxsize=2)
> def fibonacci(n: int) -> int:
>     ...
> ```
